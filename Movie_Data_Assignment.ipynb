### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import numpy as np
```

```python
genre_data = data.copy()

genre_data['Main_Genre'] = np.where(
    genre_data['Genre'].str.contains('Drama', na=False),
    'Drama',
    np.where(
        genre_data['Genre'].str.contains('Action', na=False),
        'Action',
        'Other'
    )
)

plot_data = genre_data[
    genre_data['Main_Genre'].isin(['Drama', 'Action'])
].dropna(subset=['Gross'])

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=plot_data,
    x='Main_Genre',
    y='Gross',
    hue='Main_Genre',
    palette='Set2',
    legend=False
)
plt.title('Gross Revenue: Drama vs Action')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = data.nlargest(5, 'IMDB_Rating').copy()

# IMDB rating ko 100-point scale par convert karna
top5['IMDB_Scaled'] = top5['IMDB_Rating'] * 10

# Bar chart ke liye Meta_score aur scaled rating dono available hon
comparison = top5.dropna(subset=['Meta_score', 'IMDB_Scaled'])

x = np.arange(len(comparison))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    x - width / 2,
    comparison['Meta_score'],
    width,
    label='Meta_score',
    color='steelblue'
)

plt.bar(
    x + width / 2,
    comparison['IMDB_Scaled'],
    width,
    label='IMDB Rating (out of 100)',
    color='darkorange'
)

plt.xticks(x, comparison['Series_Title'], rotation=25, ha='right')
plt.ylabel('Score out of 100')
plt.title('Top 5 Highest-Rated Movies: Meta_score vs IMDB Rating')
plt.legend()
plt.tight_layout()
plt.show()
```
