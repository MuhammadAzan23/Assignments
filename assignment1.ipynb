### Task 1 - Student Information

```python
name = "Ahmed Raza"
age = 21
city = "Karachi"
is_student = True

print("Name:", name)
print("Age:", age)
print("City:", city)
print("Is Student:", is_student)
```

### Task 2 - Data Types

```python
str_var = "Python Programming"
int_var = 42
float_var = 3.14
bool_var = True

print(str_var, type(str_var))
print(int_var, type(int_var))
print(float_var, type(float_var))
print(bool_var, type(bool_var))
```

### Task 3 - Arithmetic Operators

```python
num1 = 20
num2 = 6

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Modulus:", num1 % num2)
print("Power:", num1 ** num2)
```

### Task 4 - Student Marks Calculator

```python
sub1 = 85
sub2 = 78
sub3 = 92

total = sub1 + sub2 + sub3
average = total / 3

print("Total Marks:", total)
print("Average Marks:", average)
```

### Task 5 - Shopping Bill

```python
price = 250
qty = 4
bill = price * qty

print("Total Bill:", bill)
```

### Task 6 - Rectangle Area

```python
length = 12
width = 5
area = length * width
perimeter = 2 * (length + width)

print("Area:", area)
print("Perimeter:", perimeter)
```

### Task 7 - Time Converter

```python
total_minutes = 250
hours = total_minutes // 60
rem_minutes = total_minutes % 60

print(hours, "hours and", rem_minutes, "minutes")
```

### Task 8 - Student Eligibility

```python
marks = 65
attendance = 80

eligible = marks >= 50 and attendance >= 75
print("Eligible:", eligible)
```

### Task 9 - Logical Operators

```python
a = True
b = False

print(a and b)
print(a or b)
print(not a)
print(not b)
```
