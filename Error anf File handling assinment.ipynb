### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4 solution code here
f = open("numbers.txt", "w")
for i in range(1, 6):
    f.write(str(i) + "\n")
f.close()
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
total = 0

f = open("numbers.txt", "r")
for line in f:
    total = total + int(line.strip())
f.close()

print("Sum of numbers:", total)
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# Q7 solution code here
# creating data.txt first so we have something to read
f = open("data.txt", "w")
f.write("This is a sample data file.")
f.close()

with open("data.txt", "r") as f:
    content = f.read()

print(content)
```

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python
# Q8 solution code here
f = open("sample.txt", "w")
f.write("Line 1\nLine 2\nLine 3")
f.close()

with open("sample.txt", "r") as f:
    lines_list = f.readlines()

print(lines_list)
```

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
# Q9 solution code here
f = open("source.txt", "w")
f.write("This is the source file content.")
f.close()

with open("source.txt", "r") as f:
    data = f.read()

with open("destination.txt", "w") as f:
    f.write(data)
```

### Q10 
`os` module ka use karke check karein ki kya `test.txt` file exist karti hai ya nahi. Agar karti hai toh usay delete kar dein.

### Q13 (Easy)

### Q14

### Q15 
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:
    print("Error: cannot divide by zero.")
except ValueError:
    print("Error: invalid input.")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    num = int("25")
except ValueError:
    print("Error: invalid number.")
else:
    print("No error occurred, number is:", num)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return age

try:
    check_age(15)
except ValueError as e:
    print("Error:", e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
try:
    f = open("some_missing_file.txt", "r")
    content = f.read()
    print(content)
except FileNotFoundError:
    print("Error: file not found.")
finally:
    try:
        f.close()
    except NameError:
        pass
    print("Done.")
```
